```python
df = df[df['model_trend'] < 1]
```

```python
df['se'] = (df['model_trend'] - df['coastsat_trend']) ** 2
df.drop(columns=['model_trend', 'coastsat_trend', 'transect']).groupby('model').mean().rename(columns={'se': 'mse'})
```

```python
models = []
r2s = []
for model in df['model'].unique():
    sub = df[df['model'] == model]
    x = sub['coastsat_trend'].values
    y = sub['model_trend'].values
    a, b = np.polyfit(x, y, 1)
    yhat = x * a + b
    ss_res = ((y - yhat) ** 2).sum()
    ss_tot = ((y - y.mean()) ** 2).sum()
    r2 = 1 - ss_res / ss_tot
    models.append(model)
    r2s.append(r2)

regs = pd.DataFrame({'model': models, 'R2': r2s})
regs.to_csv('../model-trend-regs.csv')
regs
```
