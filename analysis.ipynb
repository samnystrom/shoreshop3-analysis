```python
def plot_wavelet(df: pd.DataFrame, col: str, dt: float, title: str) -> mpl.figure.Figure:
    '''
    df should have a 'date' column
    '''
    wave, scales, freqs, coi, fft, fftfreqs = pycwt.cwt(df[col].values, dt=dt, dj=1/12, s0=-1, J=-1, wavelet='morlet', freqs=None)

    power = np.abs(wave) ** 2
    fft_power = np.abs(fft) ** 2
    period = 1 / freqs
    global_power = np.mean(power, axis=1)
    t = np.arange(0, len(coi), 1)

    fig, ax = plt.subplots()

    fig.tight_layout()

    contourset = ax.contourf(t, np.log2(period), np.log2(power), levels=64, extend='both', cmap='Blues')
    ax.fill(
        np.concatenate([t, t[-1:] + dt, t[-1:] + dt, t[:1] - dt, t[:1] - dt]),
        np.concatenate([np.log2(coi), [1e-9], np.log2(period[-1:]), np.log2(period[-1:]), [1e-9]]),
        'white',
        alpha=0.5,
        #hatch='x',
    )

    def x_to_date(x, pos):
        idx = int(round(x))
        if 0 <= idx < len(df):
            return df.loc[idx, 'date'][:4]
        return ''
    ax.xaxis.set_major_formatter(mpl.ticker.FuncFormatter(x_to_date))
    #plt.setp(plt.gca().get_xticklabels(), rotation=-45, ha='left')
    ax.xaxis.set_major_locator(mpl.ticker.MultipleLocator((365*5 + 2) / dt))

    def y_to_period(y, pos):
        y = int(2**y)
        if y > 365:
            years = round(y/365)
            return f'{years} year' + ('s' if years > 1 else '')
        elif y > 30:
            months = round(y/30)
            return f'{months} month' + ('s' if months > 1 else '')
        else:
            return f'{y} day' + ('s' if y > 1 else '')
    ax.yaxis.set_major_formatter(mpl.ticker.FuncFormatter(y_to_period))

    fig.colorbar(contourset, label='Wavelet power (log2)')

    ax.set_xlim(np.min(t), np.max(t))
    ax.set_ylim(np.min(np.log2(period)), np.max(np.log2(period)))

    ax.set_ylabel('Period')
    if title:
        ax.set_title(title)

    fig.set_size_inches(8, 4)

    return fig
```

```python
start = time.perf_counter_ns()

fig = plot_wavelet(
    duck_cur[duck_cur['model'] == 'mtraboulsi689'],
    col='x1',
    dt=1.0,
    title='Power Spectrum (Morlet): Duck 1980–2023 mtraboulsi',
)
fig.set_size_inches(12, 6)
fig.savefig('figures/mtraboulsi-cur.png', dpi=180)

elapsed = time.perf_counter_ns() - start
print(f'{elapsed/1e6:.3f} ms')

fig
```
